```python
n = int(input())    #Перевод из десятичной СС в двоичную СС
stroka = str("")
while n > 0:
    m = n%2
    stroka += str(m)
    n //= 2
for i in range(1, len(stroka)+1):
    print(stroka[-i], end="")
```
