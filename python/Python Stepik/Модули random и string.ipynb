```python
import string
import random

def generate_password(length):
    password = []
    s1 = list(string.ascii_uppercase)
    s2 = list(string.ascii_lowercase)
    s3 = list(string.digits)
    s = ['l', 'I', '1', 'o', 'O', '0',]
    for i in s:
        if i in s1:
            s1.remove(i)
        if i in s2:
            s2.remove(i)
        if i in s3:
            s3.remove(i)
    spisok = s1 + s2 + s3
    for j in range(length - 3):
        password.append(random.choice(spisok))
    password.insert(random.randint(0, length - 1), random.choice(s1))
    password.insert(random.randint(0, length - 1), random.choice(s2))
    password.insert(random.randint(0, length - 1), random.choice(s3))
    print(''.join(password))
def generate_passwords(count, length):
    for i in range(count):
        generate_password(length)
n, m = int(input()), int(input())
generate_passwords(n, m)
```
