```python
from fractions import Fraction
n = int(input())
num = 0
for i in range(1, n // 2 + n % 2):
    if Fraction(i, n - i) > num and Fraction(i, n - i).numerator + Fraction(i, n - i).denominator == n:
        num = Fraction(i, n - i)
print(num)
```
