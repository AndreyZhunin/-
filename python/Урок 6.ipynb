```python
# объявление функции
def compute_binom(n, k):
    nfac = 1
    kfac = 1
    m = n - k
    mfac = 1
    for i in range(1, n + 1):
        nfac = n * i
    for j in range(1, k + 1):
        kfac = k * i
    for k in range(1, m + 1):
        mfac = m * i
    bk = nfac / (kfac * mfac)
    return int(bk)

# считываем данные
n = int(input())
k = int(input())

# вызываем функцию
print(compute_binom(n, k))
```

```python
from math import factorial
# объявление функции
def compute_binom(n, k):
    return int(factorial(n) / (factorial(k) * factorial(n - k)))

# считываем данные
n = int(input())
k = int(input())

# вызываем функцию
print(compute_binom(n, k))
```

```python
# объявление функции
def number_to_words(num):
    s = ['один', 'два', 'три', 'четыре', 'пять', 'шесть', 'семь', 'восемь', 'девять', 'десять', 'одиннадцать', 'двенадцать', 'тринадцать', 'четырнадцать', 'пятнадцать', 'шестнадцать', 'семнадцать', 'восемнадцать', 'девятнадцать', 'двадцать', 'тридцать', 'сорок', 'пятьдесят', 'шестьдесят', 'семьдесят', 'восемьдесят', 'девяносто']
    if num <= 20:
        return s[num - 1]
    elif num == 30 or num == 40 or num == 50 or num == 60 or num == 70 or num == 80 or num == 90:
        return s[num // 10 - 1 + 18]
    else:
        return s[num // 10 - 1 + 18] + ' ' + s[num % 10 - 1]
# считываем данные
n = int(input())

# вызываем функцию
print(number_to_words(n))
```

```python
# объявление функции
def is_magic(date):
    day, month, year = date.split('.')
    return int(day) * int(month) == int(year) % 100
# считываем данные
date = input()

# вызываем функцию
print(is_magic(date))
```

```python
# объявление функции
def is_pangram(text):
    s = "abcdefghijklmnopqrstuvwxyz"
    m = 0
    for i in s:
        if i not in text.lower():
            m += 1
    return m == 0
# считываем данные
text = input()

# вызываем функцию
print(is_pangram(text))
```
