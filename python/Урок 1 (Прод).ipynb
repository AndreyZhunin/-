```python
text = input() + " запретил букву"
letter = 1072
while len(text.strip()) > 0:
    counter = 0
    if chr(letter) in text:
        print(text.replace("  ", " ").strip(), " ", chr(letter), sep="")
    letter += 1
    while counter < len(text):
        text = list(text)
        if chr(letter - 1) in text:
            text.remove(chr(letter - 1))
        counter += 1
    text = "".join(text)
```

```python
def chunked(s, n):
    spisok1, spisok2 = [], []
    counter = 0
    for i in range(len(s)):
        spisok2.append(s[i])
        counter += 1
        if counter == n or i == len(s) - 1:
            spisok1.append(spisok2)
            spisok2 = []
            counter = 0
    return spisok1
s, n = input().split(), int(input())
print(chunked(s, n))
```
