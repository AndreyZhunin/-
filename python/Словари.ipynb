```python
s = list(input())    #Сообщения на кнопочном телефоне
letters = {1: '.,?!:', 2: 'ABC', 3: 'DEF', 4: 'GHI', 5: 'JKL', 6: 'MNO', 7: 'PQRS', 8: 'TUV', 9: 'WXYZ', 0: ' '}
for i in s:
    for j in letters:
        if i.upper() in letters[j]:
            print(str(j) * (letters[j].find(i.upper()) + 1), end='')
```
