```python
#Принимаем параметры матрицы, где n - количество строк, m - количество столбцов
n, m = [int(i) for i in input().split()]
#Создаем список чисел, которыми будет заполняться матрица (от 1 до n * m + 1)
spisok = [i for i in range(1, n * m + 1)]
#Создаем матрицу, состоящую из нулей
matrix = [[0] * m for _ in range(n)]
#Задаем координаты чисел в матрице, а также их коэффициеты
i, j, k, l = 0, 0, 0, 0
#Создаем цикл while, который будет повторяться, пока длина списка чисел не станет равна 0
while len(spisok) != 0:
    #Задаем условия для заполнения матрицы числами слева направо
    if i == j == 0 or matrix[i - 1][j] != 0 and matrix[i][j - 1] != 0 and matrix[i][j + 1] == 0:
        k, l = 0, 1
    #Задаем условия для заполнения матрицы числами сверху вниз
    if i == 0 and j == m - 1 or n != 2 and j < m - 1 and matrix[i][j + 1] != 0 and matrix[i - 1][j] != 0 and matrix[i + 1][j] == 0:
        k, l = 1, 0
    #Задаем условия для заполнения матрицы числами справа налево
    if i == n - 1 and j == m - 1 or i < n - 1 and matrix[i + 1][j] != 0 and matrix[i][j + 1] != 0 and matrix[i][j - 1] == 0:
        k, l = 0, -1
    #Задаем условия для заполнения матрицы числами снизу вверх
    if i == n - 1 and j == 0 and n != 1 or n != 1 and matrix[i][j - 1] != 0 and matrix[i + 1][j] != 0 and matrix[i - 1][j] == 0:
        k, l = -1, 0
    #Добавляем из списка чисел в матрицу число, которое прошло одно из предыдущих условий
    matrix[i][j] = spisok[0]
    #Удаляем это число из списка чисел
    del spisok[0]
    #При помощи коэффициентов, полученных из одного из условий выше, задаем новые координаты для следующего числа в матрице
    i += k
    j += l
#Выводим матрицу
for x in range(n):
    for y in range(m):
        print(str(matrix[x][y]).ljust(2), end=" ")
    print()
```

```python
def square_matrix_mult(matrixA, matrixB, size):
    matrixC = [[0] * size for _ in range(size)]
    for i in range(size):
        for j in range(size):
            for q in range(size):
                matrixC[i][j] += matrixA[i][q] * matrixB[q][j]
    return matrixC
    
n = int(input())
matrix = [[int(i) for i in input().split()] for _ in range(n)]
m = int(input())
powered_matrix = matrix.copy()

for _ in range(m - 1):
    powered_matrix = square_matrix_mult(matrix, powered_matrix, n)

for row in powered_matrix:
    print(*row)
```

```python
n = int(input())
matrix = [[int(i) for i in input().split()] for _ in range(n)]
summa = sum([i for i in range(1, n + 1)])
for i in range(n):
    sum_of_string, sum_of_column = 0, 0
    for j in range(n):
        sum_of_string += matrix[i][j]
        sum_of_column += matrix[j][i]
        print(sum_of_string, sum_of_column, summa)
    if sum_of_string == summa and sum_of_column == summa:
        flag = True
    else:
        flag = False
if flag == True:
    print("YES")
else:
    print("NO")
```

```python
n = list(input())    #Шахматный ферзь
text = "abcdefgh"
ij = [i for i in range(8) if text[i] == n[0]] + [int(n[1]) - 1]
matrix = [["."] * 8 for _ in range(8)]
for i in range(8):
    for j in range(8):
        if j == ij[0] or i == ij[1] or j - ij[0] == i - ij[1] or j - ij[0] == -(i - ij[1]):
            matrix[i][j] = "*"
matrix[ij[1]][ij[0]] = "Q"
for i in range(8):
    for j in range(8):
        print(matrix[-i - 1][j], end=" ")
    print()  
```

```python
n = list(input())    #Шахматный конь
text = "abcdefgh"
ij = [i for i in range(8) if text[i] == n[0]] + [int(n[1]) - 1]
matrix = [["."] * 8 for _ in range(8)]
matrix[ij[1]][ij[0]] = "N"
if ij[0] < 7 and ij[1] < 6:
    matrix[ij[1] + 2][ij[0] + 1] = "*"
if ij[0] > 0 and ij[1] < 6:
    matrix[ij[1] + 2][ij[0] - 1] = "*"
if ij[0] < 6 and ij[1] < 7:
    matrix[ij[1] + 1][ij[0] + 2] = "*"
if ij[0] < 6 and ij[1] > 0:
    matrix[ij[1] - 1][ij[0] + 2] = "*"
if ij[0] < 7 and ij[1] > 1:
    matrix[ij[1] - 2][ij[0] + 1] = "*"
if ij[0] > 0 and ij[1] > 1:
    matrix[ij[1] - 2][ij[0] - 1] = "*"
if ij[0] > 1 and ij[1] > 0:
    matrix[ij[1] - 1][ij[0] - 2] = "*"
if ij[0] > 1 and ij[1] < 7:
    matrix[ij[1] + 1][ij[0] - 2] = "*"
for i in range(8):
    for j in range(8):
        print(matrix[-i - 1][j], end=" ")
    print()    
```

```python
n = int(input())
matrix = [[0] * n for _ in range(n)]
for i in range(n):
    for j in range(n):
        if i == j + (i - j):
            matrix[i][j] = abs(i - j)
        print(str(matrix[i][j]).ljust(2), end=" ")
    print()
```
